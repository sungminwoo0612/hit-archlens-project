## 4. 데이터셋 확인

1. 데이터셋 준비: `python scripts/prepare_yolo_dataset.py --mode fine`
2. 모델 학습: `python scripts/train_yolo_cls.py --mode fine --epochs 100`
3. 모델 평가: `python scripts/eval_yolo_cls.py --model runs/classify/fine_cls_*/weights/best.pt`
4. 이미지 예측: `python scripts/predict_yolo_cls.py --model runs/classify/fine_cls_*/weights/best.pt --source dataset/icons/images/...`

```python
from pathlib import Path

data_dir = Path.cwd() / "dataset" / "icons"
print("[3/3] 데이터셋 확인 중...")

if not data_dir.is_dir():
    print(f"[WARN] 데이터셋 디렉터리가 없습니다: {data_dir}")
else:
    print(f"[OK] 데이터셋 디렉터리: {data_dir}")
    for mode in ["fine", "coarse"]:
        subdir = data_dir / f"yolo_cls_{mode}"
        if subdir.is_dir():
            print(f"[OK] {mode} 데이터셋 준비됨")
        else:
            print(f"[WARN] {mode} 데이터셋이 없습니다. 다음 명령을 실행하세요:")
            print(f"       python scripts/prepare_yolo_dataset.py --mode {mode}")
```
